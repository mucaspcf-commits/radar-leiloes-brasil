# Qualidade da lista Caixa
Unidade: imóvel por código da fonte. Perfil reproduzível, sem inferência de disponibilidade ou rentabilidade. Fonte: https://venda-imoveis.caixa.gov.br/listaweb/Lista_imoveis_geral.csv
Execute na raiz do projeto. Preços acima da avaliação são preservados para revisão.

```python
import json
from pathlib import Path
data = json.loads(Path("static/data.json").read_text(encoding="utf-8"))
rows = data["properties"]
profile = dict(registros=len(rows), ids_unicos=len({r["id"] for r in rows}), ufs=len({r["uf"] for r in rows}), cidades=len({(r["uf"],r["cidade"]) for r in rows}), areas_ausentes=sum(r.get("area") is None for r in rows), acima_avaliacao=sum(r["preco"] > r["avaliacao"] for r in rows), data_fonte=data["source_date"])
assert profile["registros"] == profile["ids_unicos"]
print(json.dumps(profile, ensure_ascii=False, indent=2))
```
